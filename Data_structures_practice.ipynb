Problem statement:

Given a list of integers, create a dictionary where:

keys are the unique elements

values are the number of times each element appears

Input
nums = [1, 2, 2, 3, 1, 4, 2]

{1: 2, 2: 3, 3: 1, 4: 1}

```python
nums = [1, 2, 2, 3, 1, 4, 2] #here iam taking the input
freq={} #here iam just taking this empty i will just add the 
for i in nums:
    if i not in freq:
        freq[i]=1 #here iam just adding this key and value if we dont have this in frq we are not able to update the value if we dont have this in freq already we will get key error
    else:
        freq[i]+=1 #here iam just updating the key's value by adding 1 if the key is already in this freq
print(freq) #here iam just printing the new updates feq of all the number i took
```

QUESTION 2 — Find First Non-Repeating Element

Problem statement:

Given a list of integers, find the first element that appears only once.

Input
nums = [2, 3, 4, 2, 3, 5, 4, 6]

Expected Output
5


```python
nums = [2, 3, 4, 2, 3, 5, 4, 6]
freq={}
for i in nums:
        if i not in freq:
            freq[i]=1
        else:
            freq[i]+=1
print(freq)

for i in nums:
        if freq[i]==1:
            print(i)
            break
```

                                    In-place Reverse
In this we are not going to take an extra list to reverse a list we are goin to use swapping with indexes

```python
arr = [10, 20, 30, 40, 50]
#rev=[] This time iam not going to use this empty list to append this rev for each and evry iteration
left=0#here iam taking indexes this left is start index
right=len(arr)-1 #This right is end index
while left<right:
    arr[left],arr[right]=arr[right],arr[left] #here iam swapping the right with left and left with right using indexing
    left+=1 #from here adding 1 to left so then after index 0 then i will go for 1,2..
    right-=1 #iam decreasing the index from right

print(arr)
```
